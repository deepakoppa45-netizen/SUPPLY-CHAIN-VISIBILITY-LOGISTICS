```python
#Calculate the correct delay
df["Delay_Days"] = (
    df["Delivery_Date"] - df["EDD"]
).dt.days
df['Delay_Days'].describe()
```

```python
#Recreate the Status column
df.loc[df["Delay_Days"] < 0, "Status"] = "Early"

df.loc[df["Delay_Days"] == 0, "Status"] = "On Time"

df.loc[df["Delay_Days"] > 0, "Status"] = "Delayed"
df[['Status']]
```

```python
#fill Fill missing Order_Date
#check how many days usually pass between ordering and dispatch.
processing_days = (
    df["Dispatch_Date"] - df["Order_Date"]
).dt.days

print(processing_days.value_counts().sort_index())
```

```python
#If most orders are dispatched after 1 day? answer-find mode
processing_mode = processing_days.mode()[0]

print(processing_mode)
```

```python
#Fill missing Order_Date
mask = df["Order_Date"].isna()

df.loc[mask, "Order_Date"] = (
    df.loc[mask, "Dispatch_Date"] - pd.to_timedelta(processing_mode, unit="D")
)
df['Order_Date'].isnull().sum()
```

```python
#Fill missing Dispatch_Date
mask = df["Dispatch_Date"].isna()

df.loc[mask, "Dispatch_Date"] = (
    df.loc[mask, "Order_Date"] + pd.to_timedelta(processing_mode, unit="D")
)
df['Dispatch_Date'].isnull().sum()
```

```python
#Expected Delivery time
expected_days = (
    df["EDD"] - df["Order_Date"]
).dt.days

print(expected_days.value_counts().sort_index())
```

```python
#Find the mode-doesn't have a single expected delivery time

expected_mode = expected_days.mode()[0]
print(expected_mode)
```

```python
#Fill missing EDD
mask = df["EDD"].isna()

df.loc[mask, "EDD"] = (
    df.loc[mask, "Order_Date"] + pd.to_timedelta(expected_mode, unit="D")
)
df['EDD'].isnull().sum()
```

```python
#Recalculate Delay_Days again
df["Delay_Days"] = (
    df["Delivery_Date"] - df["EDD"]
).dt.days
df['Delay_Days']
```

```python
#Recreate Status
df.loc[df["Delay_Days"] < 0, "Status"] = "Early"

df.loc[df["Delay_Days"] == 0, "Status"] = "On Time"

df.loc[df["Delay_Days"] > 0, "Status"] = "Delayed"
df['Status']
```

```python
df[[
    "Order_Date",
    "Dispatch_Date",
    "EDD",
    "Delivery_Date",
    "Delay_Days",
    "Status"
]].isnull().sum()
```

```python
#fill Tran-Cost	and Order_Value
df['Tran-Cost']=df['Tran-Cost'].fillna(df['Tran-Cost'].median())

df['Tran-Cost'].isnull().sum()
```

```python

df['Order_Value']=df['Order_Value'].fillna(df['Order_Value'].median())
df['Order_Value'].isnull().sum()
```
